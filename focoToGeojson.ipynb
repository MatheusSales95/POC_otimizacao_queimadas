```python
import geopandas  as gpd
import pandas as pd
```

```python
focos_csv = gpd.read_file('bdqueimadas_2024-08-04_2024-09-04.csv')
```

```python
focos_csv.head()
```

```python
focos_csv.drop(columns=['Bioma', 'Precipitacao', 'DiaSemChuva', 'RiscoFogo', 'FRP', 'Estado', 'Pais', 'Municipio'], inplace=True)
```

```python
geometry = gpd.points_from_xy(focos_csv['Longitude'], focos_csv['Latitude'])
gdf = gpd.GeoDataFrame(focos_csv, geometry=geometry, crs='EPSG:4326')
```

```python
gdf['id_foco'] = gdf.index
```

```python
gdf.rename(columns={'id_foco': 'id','DataHora':'data', 'Satelite': 'satelite'}, inplace=True)
```

```python
if not pd.api.types.is_datetime64_any_dtype(gdf["data"]):
        gdf["data"] = pd.to_datetime(gdf["data"])
```

```python
gdf.drop(columns=['Latitude', 'Longitude'], inplace=True)
```
